```python
import os
from obspy.core import read


dir_path = './'
save_path = './text_data'

for file_name in os.listdir(dir_path):
    if os.path.isfile(os.path.join(dir_path, file_name)) and os.path.join(dir_path, file_name).format().endswith("BHZ"):

        track = read(os.path.join(dir_path, file_name))
        # print(track[0].data)
        with open(os.path.join(save_path, file_name.replace(".BHZ", ".txt")), 'w') as f:
            for nth,data in enumerate(track[0].data):
                f.write(str(data))
                if(nth+1 != len(track[0].data)):
                    f.write(' ')

```
