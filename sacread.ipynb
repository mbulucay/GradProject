```python
from obspy import read

filenames = ['CAVI.BHZ', 'BAKL.BHZ', 'KULA.BHZ', 'SUSR.SHZ']


st = read(filenames[1] + '.sac', debug_headers=True)
print(st)
```

```python
print(st[0].stats)
```

```python
# plot settings
print(st[0].stats.starttime + 20)
print(st[0].stats.endtime)

```
