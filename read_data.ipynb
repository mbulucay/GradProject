```python
from obspy.core import read
singlechannel = read('./3514_tdvms_N_noise.mseed')
print(singlechannel)
singlechannel.plot()


for i,tr in enumerate(singlechannel):
    # save the trace to a file
    with open(f"./text_data/noise{i}.txt", 'w') as f:
        for i, data in enumerate(tr.data):
            f.write(str(data))
            if(i != len(tr.data)-1):
                f.write(" ")
```
