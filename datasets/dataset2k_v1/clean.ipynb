```python

import os
from obspy import read

clean_class_folder = "class0/clean/"
clean_trace_file_list = []


for trace_file in os.listdir(clean_class_folder):
    clean_trace_file_list.append(str(trace_file))



original_class_folder = "class0/original/"
original_trace_file_list = []

for trace_file in os.listdir(original_class_folder):
    original_trace_file_list.append(str(trace_file))

original_trace_file_list.sort()
clean_trace_file_list.sort()
```

```python
org = read(original_class_folder + original_trace_file_list[0])[0]
clean = read(clean_class_folder + clean_trace_file_list[0])[0]

print(original_trace_file_list[80])
org.plot()
print(clean_trace_file_list[80])
clean.plot()
```

```python
from obspy import read

# plot the original and clean traces
for i, trace in enumerate(clean_trace_file_list):
    print(f"clean trace: {i}: ", trace)
    read_clean_trace = read(clean_class_folder + trace)[0]
    read_clean_trace.plot()
```

```python
from obspy import read

# plot the original and clean traces
for i, trace in enumerate(original_trace_file_list):
    print(f"original trace: {i} ", trace)
    read_clean_trace = read(original_class_folder + trace)[0]
    read_clean_trace.plot()
```
